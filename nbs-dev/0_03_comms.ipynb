## Proxy and VPN

```python
#| export
def clean_no_proxy_12vpn():
    """Remove tokens like [fec0::], [fc00::], [fe80::]

    12VPN sets NO_PROXY to include tokens which causes problems with some libraries (e.g. openai, ...) that don't handle them correctly. 
    This function cleans the NO_PROXY environment variable by removing such tokens.
    """
    pattern = re.compile(r'^\[[^,\]]*::\*\]$')
    raw = os.environ.get("NO_PROXY") or os.environ.get("no_proxy") or ""
    items = [x.strip() for x in raw.split(",") if x.strip()]

    cleaned_items = [x for x in items if not pattern.match(x)]
    cleaned = ",".join(cleaned_items)

    os.environ["NO_PROXY"] = cleaned
    os.environ["no_proxy"] = cleaned
    return cleaned
```
